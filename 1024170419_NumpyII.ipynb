#1. Consider the following sensor readings: temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40]) A sensor has an average measurement error of +2°C.

**a. Add 2°C to every reading using vectorization.**

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2

print(corrected_temperature)
```

**b. Convert all temperatures from Celsius to Fahrenheit using: F = (9/5)C + 32**

```python
fahrenheit = (9/5) * corrected_temperature + 32

print(fahrenheit)
```

**c. Identify all readings greater than 32°C using Boolean indexing.**

```python
greater_than_32 = corrected_temperature[corrected_temperature > 32]

print(greater_than_32)
```

**d. Count how many readings exceed 32°C.**

```python
count = np.sum(corrected_temperature > 32)
print(count)
```

**e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.**

vectorization allows NumPy to perform operations on entire arrays at once instead of processing each element individually using a loop. This makes calculations faster.

   Boolean indexing allows filtering data directly based on conditions without writing loops and manually checking each value. It makes the code shorter, easier to read, and more efficient when working with large datasets.

#Q2. The following matrix represents the number of daily steps recorded by four users over three days:

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```

a. Calculate the total steps recorded.d

```python
total_steps = np.sum(steps)
print(total_steps)
```

b. Calculate the mean number of steps.

```python
mean_steps = np.mean(steps)
print(mean_steps)
```

c. Find the maximum and minimum values.

```python
maximum = np.max(steps)
minimum = np.min(steps)
print("Maximum:", maximum)
print("Minimum:", minimum)
```

d. Calculate the total steps for each day using axis=0.

```python
daily_total = np.sum(steps, axis=0)
print(daily_total)
```

e. Calculate the total steps for each user using axis=1.

```python
user_total = np.sum(steps, axis=1)
print(user_total)
```

f. Find the user/day position containing the maximum number of steps using argmax().

```python
position = np.unravel_index(np.argmax(steps), steps.shape)
print(position)
```

Q4. A simple cognitive-assistive system wants to predict a user's assistance score based on three factors:
Sleep hours
Activity level
Stress level

```python
y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
```

a. Display the shape and dimensions of X.

```python
print("Shape:", X.shape)
print("Dimensions:", X.ndim)
```

. Calculate: X.T and explain what the transpose represent

```python
print(X.T)
```

c. Calculate the matrix product: X.T @ X

```python
XT_X = X.T @ X

print(XT_X)
```

d. Calculate the matrix inverse using: np.linalg.inv() where appropriate.

```python
inverse = np.linalg.inv(XT_X)
print(inverse)
```

e. Implement the Ordinary Least Squares equation: 𝛽 = (𝑋ᵀ𝑋)⁻¹𝑋ᵀ𝑦 using NumPy.

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print(beta)
```

. Explain what the resulting coefficients represent in the context of the three user features.

The three coefficients correspond to:
Sleep hours       → -0.15921454
Activity level    →  0.14083853
Stress level      → 10.06302242


g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to obtain a predicted assistance score.

```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
