#3. Set the random seed to your own roll number

```python
import random

random.seed(1024170419)

numbers = [random.randint(100, 900) for i in range(100)]

print(numbers)
```

**3(i). Generate a list of 100 random numbers between 100 and 900 (inclusive). Count and print all odd numbers in the list.**

```python
odd_numbers = [x for x in numbers if x % 2 != 0]

print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

**3(ii). Count and print all even numbers in the list.**

```python
even_numbers = [x for x in numbers if x % 2 == 0]

print("Even numbers-->", even_numbers)
print("Count of even numbers-->", len(even_numbers))
```

**3(iii). Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.**

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
prime= [x for x in numbers if is_prime(x)]
print("Prime numbers:", prime)
print("Count of prime numbers:", len(prime))
```

**3(iv). Print the number that occurs most frequently in your list, and how many times it occurs.**

```python
frequency = {}

for x in numbers:
    frequency[x] = numbers.count(x)

most_frequent = max(frequency, key=frequency.get)

print("Most frequent=", most_frequent)
print("Occurrence=", frequency[most_frequent])
```

**4 Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.**

```python
digits = [1, 0, 2, 4, 1, 7, 0, 4]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A=", A)
print("Set B=", B)
```

**4(vi). Find the union of A and B — all unique values across both sets.**

```python
union_set = A.union(B)

print("Union:",union_set)
```

**4(vii). Find the intersection of A and B — values common to both.**

```python
intersection_set = A.intersection(B)

print("Intersection:",
      intersection_set)
```

**4(viii). Find A − B and B − A separately using difference(), and explain in one line how this differs from symmetric_difference().**

```python
A_diff_B = A.difference(B)
B_diff_A = B.difference(A)
print("A -B:", A_diff_B)
print("B -A:", B_diff_A)
```

difference() gives values present in one set but not the other. symmetric_difference() gives all values that are unique to either set.

**4(ix). Find the symmetric difference of A and B.**

```python
symmetric_difference = A.symmetric_difference(B)

print("Symmetric Difference:",symmetric_difference)
```

**4(x). Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().**

```python
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

**4(xi). Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists.**

```python
X = int(input("Enter value to remove from A: "))
A.discard(X)
print("Set A after removal:", A)
```

discard() is safer than remove() because it does not give an error if the element is not present in the set.
